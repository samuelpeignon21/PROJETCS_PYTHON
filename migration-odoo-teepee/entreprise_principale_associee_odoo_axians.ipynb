```python
#lire le fichier csv CONTACT ODOO
df_Mapping = pd.read_excel('/content/Copie de Mapping sites ↔ entreprisest.xlsx')
#Afficher tout le df
pd.set_option('display.max_rows', None)
```

```python
df_Mapping
```

```python
df_Mapping2 = df_Mapping.copy()
```

```python
df_Mapping2["Teepee -> Nom de l'entreprise"] = (
    df_Mapping["Teepee -> Nom de l'entreprise"]
    .fillna(df_Mapping["Teepee -> Nom du site"])
)
```

```python
df_Mapping2["Teepee -> Nom de l'entreprise"] = (
    df_Mapping2["Teepee -> Nom de l'entreprise"]
    .str.strip()
    .str.upper()
)
```

```python
df_Mapping2["Teepee -> Nom du site"] = df_Mapping2["Teepee -> Nom du site"].str.strip().str.upper()
```

```python
df_Mapping2
```

```python
df_entreprises = (
    df_Mapping2[["Teepee -> Nom de l'entreprise"]]
    .drop_duplicates()
    .reset_index(drop=True)
)
```

```python
# Dico du Site -> Entreprise
dict_site_entreprise = (
    df_Mapping2
    .set_index("Teepee -> Nom du site")["Teepee -> Nom de l'entreprise"]
    .to_dict()
)
```

```python
dict_site_entreprise
```

```python
#lire le fichier csv SITES Teepee
df_Sites = pd.read_excel('/content/Clients_Export.xlsx')
#Afficher tout le df
pd.set_option('display.max_rows', None)
```

```python
df_Sites.columns
```

```python
df_Sites["Nom Entreprise (Nom)"] = df_Sites["Nom Entreprise (Nom)"].str.strip().str.upper()
```

```python
df_Sites["Entreprise"] = df_Sites["Nom Entreprise (Nom)"].map(dict_site_entreprise)
```

```python
df_Sites = df_Sites.dropna(subset=["Entreprise"]).reset_index(drop=True)
```

```python
# definitions d'une fonction
from rapidfuzz import fuzz
import pandas as pd

def sont_doublons(a, b, seuil=90):
    if pd.isna(a) or pd.isna(b):
        return False
    return fuzz.token_sort_ratio(a.lower(), b.lower()) >= seuil
```

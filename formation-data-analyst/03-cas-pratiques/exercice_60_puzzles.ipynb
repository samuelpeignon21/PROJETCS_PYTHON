Nous sommes enfin prêts à créer un graphique en chandeliers, qui est un outil très courant utilisé pour analyser les données relatives au cours des actions.  Un graphique en chandelier montre le cours d'ouverture, le cours de clôture, le cours le plus élevé et le cours le plus bas d'une action au cours d'une période donnée.  La couleur de la « bougie » (la partie épaisse de la barre) est verte si l'action a clôturé au-dessus de son cours d'ouverture, ou rouge si elle a clôturé en dessous.

![Candlestick Example](img/candle.jpg)

Cet exemple a été initialement conçu pour être un défi de tracé pandas, mais il se trouve que ce type de tracé n'est tout simplement pas réalisable avec les méthodes pandas.  Si vous n'êtes pas familier avec matplotlib, nous avons fourni une fonction qui tracera le graphique pour vous à condition que vous puissiez utiliser pandas pour obtenir les données dans le bon format.

La première étape consiste à mettre les données dans le bon format à l'aide de la fonction de regroupement des séries temporelles de pandas.  Nous souhaitons que chaque bougie représente une heure de données.  Vous pouvez écrire votre propre fonction d'agrégation qui renvoie l'ouverture/le haut/le bas/la fermeture, mais pandas dispose d'une fonction intégrée qui permet également de le faire.

La cellule ci-dessous contient des fonctions d'aide.  Appelez ```day_stock_data()``` pour générer un DataFrame contenant les prix de vente d'une action hypothétique, et l'heure à laquelle la vente a eu lieu.  Appelez ```plot_candlestick(df)``` sur vos données boursières correctement agrégées et formatées pour imprimer le graphique en chandeliers.

````python
import numpy as np
def float_to_time(x):
    return str(int(x)) + ":" + str(int(x%1 * 60)).zfill(2) + ":" + str(int(x*60 % 1 * 60)).zfill(2)

def day_stock_data():
    #NYSE is open from 9:30 to 4:00
    time = 9.5
    price = 100
    results = [(float_to_time(time), price)]
    while time < 16:
        elapsed = np.random.exponential(.001)
        time += elapsed
        if time > 16:
            break
        price_diff = np.random.uniform(.999, 1.001)
        price *= price_diff
        results.append((float_to_time(time), price))


    df = pd.DataFrame(results, columns = ['time','price'])
    df.time = pd.to_datetime(df.time)
    return df

#Don't read me unless you get stuck!
def plot_candlestick(agg):
    """
    agg is a DataFrame which has a DatetimeIndex and five columns: ["open","high","low","close","color"]
    """
    fig, ax = plt.subplots()
    for time in agg.index:
        ax.plot([time.hour] * 2, agg.loc[time, ["high","low"]].values, color = "black")
        ax.plot([time.hour] * 2, agg.loc[time, ["open","close"]].values, color = agg.loc[time, "color"], linewidth = 10)

    ax.set_xlim((8,16))
    ax.set_ylabel("Price")
    ax.set_xlabel("Hour")
    ax.set_title("OHLC of Stock Value During Trading Day")
    plt.show()
````

- #### Créez un super-dataset !
Pour cela, sélectionnez toutes les lignes de la table Quality, ajoutez les informations "BRAND" et "LABEL_PRODUCT" depuis la table Product. Sélectionnez les valeurs de ce dataset où les marques (brand) ne sont pas null. Ordonnez la date par ordre ascendant et veuillez renseigner la première observation qui apparaît.

````python
# Mets ton code SQL entre les """, à la place des "..."

df_sql = pd.read_sql(

    """
Select *
from quality as q
left join products as p
on PRODUCT_ID = ID
where brand is not null
Order by order_date ASC
limit 1;

    """

    , con=engine)

# Afficher le résultat
display(df_sql)
````

- #### Comptez les marques qui ont été vendues plus d'une fois par le marchand n° 332.

````python
# Mets ton code SQL entre les """, à la place des "..."

df_sql = pd.read_sql(

    """
WITH test AS (
    SELECT *
    FROM quality AS q
    LEFT JOIN products AS p
    ON q.PRODUCT_ID = p.ID
    WHERE brand IS NOT NULL
    ORDER BY order_date ASC
),
test_2 AS (
    SELECT brand, COUNT(brand) AS nbr_marques_vendu
    FROM test
    WHERE merchant_id = 332
  	group by brand
)

SELECT *
FROM test_2
where nbr_marques_vendu > 1;

    """

    , con=engine)

# Afficher le résultat
display(df_sql)
````

#### Questions sur les statistiques inférentielles :
Testez une hypothèse pour comparer le nombre moyen de vaccinations quotidiennes par million (`daily_vaccinations_per_million`) entre deux périodes différentes : du 01/01/2021 au 31/01/2021 (correspondant au mois de janvier), et du 01/02/2021 au 28/02/2021 (correspondant au mois de février), (supposons que les échantillons partagent la même taille).


#### Question : Formulez H0 et H1

H0 : Il n’y a pas de différence entre les moyennes de daily_vaccinations_per_million en janvier et en février

H1 : Il existe une différence entre les moyennes.


#### Question : Y a t-il un différence significative au seuil de 5% entre les deux moyennes, rejetons-nous H0 ? Interpréter le résultat du t-test en automatisant la réponse à l'aide d'une condition (*si p-value > ... alors ... sinon ...*).


Note :
- Mettre le paramètre **`equal_var=`** suivi de l'argument **`False`** pour adapter votre test t de student.
- Pensez aux dates

````python
janvier.columns
````

````python
janvier = df_test[df_test["date"].str.contains("2021-01")]
fevrier = df_test[df_test["date"].str.contains("2021-02")]
janvier = janvier["daily_vaccinations_per_million"]
fevrier = fevrier["daily_vaccinations_per_million"]
````

````python
stat, p_value = ttest_ind(janvier, fevrier, equal_var=False)

# Résultat
print(f"Statistique t : {stat:.2f}")
print(f"P-value : {p_value:.4f}")

alpha = 0.05
if p_value < alpha:
    print("→ On rejette H0 : les moyennes sont significativement différentes.")
else:
    print("→ On ne rejette pas H0 : pas de différence significative entre les moyennes.")
````

15. Modifie la fonction `display_nb_apparitions_by_character()` afin qu'elle retourne une liste de 20 dictionnaires (un pour chaque personnage).  

**Astuce**: Chaque dictionnaire devrait contenir 3 clés : `id`, `name` et `nb_episodes`.
Nomme cette nouvelle fonction `get_nb_apparitions_by_character()`.

```python
def get_nb_apparitions_by_character(data):
  Liste = []
  for x in range(len(data["results"])):
    ID = data["results"][x]["id"]
    Name = data["results"][x]["name"]
    nb_epi = len(data["results"][x]["episode"])
    dictionnaire = {"id":ID, "name":Name, "nb_episodes":nb_epi}
    Liste.append(dictionnaire)
  return(Liste)
```

```python
get_nb_apparitions_by_character(data)
```

Enfin, joins les deux DataFrames ensemble.

```python
def get_nb_apparitions_by_character(data):
  Liste = []
  for x in range(len(data["results"])):
    ID = data["results"][x]["id"]
    Name = data["results"][x]["name"]
    nb_epi = len(data["results"][x]["episode"])
    planète = data["results"][x]["origin"]["name"]
    dictionnaire = {"id":ID, "name":Name, "nb_episodes":nb_epi, "planete" : planète}
    Liste.append(dictionnaire)
  return(Liste)
```

```python
get_nb_apparitions_by_character(data)
```

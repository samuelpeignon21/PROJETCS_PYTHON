*(Optionnel : Utilise la fonction `plot_q8()` pour visualiser la relation entre l'approvisionnement en énergie par habitant VS les documents exploitables par habitant)*

````python
def plot_q8(df):
    import matplotlib as plt  # Importe la bibliothèque matplotlib pour créer des graphiques.
    %matplotlib inline

    Top15 = df.copy()  # Crée une copie du DataFrame original pour éviter de modifier les données originales.

    # Calcule l'estimation de la population ('PopEst') pour chaque pays :
    Top15['PopEst'] = Top15['Energy Supply'] / Top15['Energy Supply per Capita']

    # Calcule le nombre de documents citables par habitant ('Citable docs per Capita') pour chaque pays :
    Top15['Citable docs per Capita'] = Top15['Citable documents'] / Top15['PopEst']

    # Crée un graphique en nuage de points (scatter plot) avec les spécifications suivantes :
    Top15.plot(
        x='Citable docs per Capita',  # Utilise la colonne 'Citable docs per Capita' pour l'axe des x.
        y='Energy Supply per Capita',  # Utilise la colonne 'Energy Supply per Capita' pour l'axe des y.
        kind='scatter',  # Spécifie que le type de graphique est un scatter plot.
        xlim=[0, 0.0001]  # Définis la limite de l'axe x pour zoomer sur une plage spécifique de valeurs.
    )
````

#### Question Optionnel

Utilise la fonction `plot_optional()` pour voir un exemple de visualisation.

````python
def plot_optional(df):
    import matplotlib as plt  # Importe la bibliothèque matplotlib pour la création de graphiques.
    %matplotlib inline

    Top15 = df.copy()  # Crée une copie du DataFrame original pour ne pas le modifier directement.

    # Crée un graphique en nuage de points (scatter plot) avec les spécifications suivantes :
    ax = Top15.plot(
        x='Rank',  # Utilise la colonne 'Rank' pour l'axe des x.
        y='% Renewable',  # Utilise la colonne '% Renewable' pour l'axe des y.
        kind='scatter',  # Spécifie que le type de graphique est un scatter plot.
        c=['#e41a1c','#377eb8','#e41a1c','#4daf4a','#4daf4a','#377eb8','#4daf4a','#e41a1c',
            '#4daf4a','#e41a1c','#4daf4a','#4daf4a','#e41a1c','#dede00','#ff7f00'],  # Définit une liste de couleurs pour chaque point, représentant les continents.
        xticks=range(1,16),  # Définit les graduations de l'axe x de 1 à 15.
        s=6*Top15['2014']/10**10,  # La taille des bulles est proportionnelle au PIB de 2014 (colonne '2014'), mis à l'échelle.
        alpha=.75,  # Rend les bulles semi-transparentes.
        figsize=[16,6]  # Définit la taille du graphique.
    )

    # Annotate (ajoute des labels) pour chaque point sur le graphique :
    for i, txt in enumerate(Top15.index):  # Parcourt chaque index du DataFrame.
        ax.annotate(txt, [Top15['Rank'][i], Top15['% Renewable'][i]], ha='center')  # Ajoute le nom du pays à chaque point.

    # Affiche un texte explicatif pour l'utilisateur :
    print("Voici un exemple de visualisation qui peut être créée pour aider à comprendre les données. "
          "C'est un graphique à bulles montrant % Renewable vs. Rank. La taille des bulles correspond au PIB de 2014 "
          "et la couleur correspond au continent.")
````

````python
import pandas as pd

# preprocessing: pivot table
hi = pd.read_csv('honolulu_trade.csv', header=3)
hi = hi.drop(columns=['Port', 'Unnamed: 5'])
hi.columns = ['year', 'country', 'commodity', 'value']
hi.value = hi.value.apply(lambda x: int(x.replace(',', ''))/1000) # la valeur sera en milliers de $
hi = pd.pivot_table(hi, values='value', columns='year', index=['country', 'commodity'])
hi = hi.sort_values(by=['country', 'commodity'])
hi = hi.reset_index()
hi
````

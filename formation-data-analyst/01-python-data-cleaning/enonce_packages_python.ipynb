### Exercice 3** - Pas de négativité par ici !

**Remplacer les valeurs négatives d'un array par la valeur 10.**

> _Bonus : Essayer de le faire sans boucle._

```python
# Reprenons l'array de l'exercice.

test_x = np.array([-1, -4, 0, 2, 3, 4, 5, -6])

print(test_x)
```

```python
test_x = np.array([-1, -4, 0, 2, 3, 4, 5, -6])
# Utilisation de np.where() pour remplacer les valeurs négatives par 10
test_x = np.where(test_x < 0, 10, test_x)
print(test_x)
```

```python
List = []
for x in test_x:
  if x < 0:
    List.append(10)
  else:
    List.append(x)
print(List)
```

### Exercice 3** – Nettoyer un DataFrame

La situation est courante : vous recevez des données contenant du texte, des listes, des chiffres mal formatés, le tout avec des valeurs manquantes. Comment pourriez-vous nettoyer ces données pour conduire votre analyse ?

Nous utiliserons la DataFrame ci-dessous :

```
            From_To  FlightNumber  RecentDelays              Airline
0      LoNDon_paris       10045.0      [23, 47]               KLM(!)
1      MAdrid_miLAN           NaN            []    <Air France> (12)
2  londON_StockhOlm       10065.0  [24, 43, 87]  (British Airways. )
3    Budapest_PaRis           NaN          [13]       12. Air France
4   Brussels_londOn       10085.0      [67, 32]          "Swiss Air"
```

Exécutez la première cellule pour créer le DataFrame, puis suivez les questions !

````python
df = pd.DataFrame({'From_To': ['LoNDon_paris', 'MAdrid_miLAN', 'londON_StockhOlm',
                               'Budapest_PaRis', 'Brussels_londOn'],
              'FlightNumber': [10045, np.nan, 10065, np.nan, 10085],
              'RecentDelays': [[23, 47], [], [24, 43, 87], [13], [67, 32]],
                   'Airline': ['KLM(!)', '<Air France> (12)', '(British Airways. )',
                               '12. Air France', '"Swiss Air"']})

df
````

**1.** Certaines valeurs sont manquantes (`NaN`) dans la colonne **FlightNumber**. Les numéros de vols devraient augmenter de 10 en 10 à chaque ligne : modifier `df` pour remplir les NaNs avec les valeurs adéquates. Puis, convertir cette colonne en type `int` (et non `float` comme actuellement). (*, @)

_Indice : L'opération pour remplacer les NaNs est ici l'interpolation._

````python
df.shape
````

````python
df.info()
````

````python
df['FlightNumber'].fillna(0, inplace=True)
````

````python
df.loc[1, 'FlightNumber'] = 10055
df.loc[3, 'FlightNumber'] = 10075
df
````

````python
df['FlightNumber'].interpolate()
````

````python
df["FlightNumber"] = df['FlightNumber'].astype(int)
df
````

````python
df.info()
df
````

**2.** Il serait plus pratique que la colonne **From_To** donne lieu à 2 colonnes séparées ! Séparer chaque string au niveau du délimiteur `_` pour créer une nouvelle DataFrame, appelée `temp`, qui contient les 2 nouvelles colonnes. Nommer ces colonnes 'From' et 'To'. (**, @)

_Indice : il s'agit ici d'appliquer une string method à toute une colonne._

````python
temp = df['From_To'].str.split('_', expand=True)
temp.columns = ['From', 'To']
temp
````

````python
temp.rename(columns={0:'From', 1:'To'}, inplace=True)
temp
````

**3.** On remarque que les lettres majuscules et minuscules sont mal utilisées dans ce DataFrame. Standardiser les strings afin que seule la première lettre soit en majuscule. (*)

````python
temp['From'] = temp['From'].str.capitalize()
temp['To'] = temp['To'].str.capitalize()
temp
````

**4.** Réinsérer le DataFrame temporaire dans notre DataFrame de base, puis supprimer la colonne **From_To**. (**)

````python
df = temp.join(df)
df.drop('From_To', axis=1, inplace=True)
df
````

**5**. Dans la colonne **Airline** on observe de la ponctuation et des symboles inutiles autour des noms de compagnies aériennes. Modifier la colonne pour ne conserver que les noms de compagnies : `'<Air France> (12)'` devient `'AirFrance'`. (***)

_Indice : se souvenir de la string method `.isalpha()`. Une bonne réponse pourrait être une fonction que l'on applique à toute la colonne à l'aide de `apply`._

````python
df['Airline']
````

````python
df['Airline'].apply(lambda x : [char for char in x if char.isalpha()]) ## corrige mais en liste chelou
````

````python
df['Airline'] = df['Airline'].apply(lambda x : "".join(char for char in x if char.isalpha()))
df
````

````python
df['Airline'].apply(lambda x : for y in x if y.isalpha() == True else '') ## une erreur
````

**6**. Pour terminer, dans la colonne **RecentDelays**, les valeurs ont été entrées en tant que liste. On veut que chaque valeur soit dans sa propre colonne, et s'il n'y a pas de nième valeur, qu'on trouve un NaN. (***, @)

_Indice : pour "étendre" la colonne `RecentDelays`, lui appliquer la fonction `pd.Series`, avec `apply`. On obtient un DataFrame temporaire que vous pouvez appeler `delays`. Renommez ses colonnes `delay_1` , `delay_2`, etc., puis réinsérez-le dans le DataFrame de base._

````python
delays = df["RecentDelays"].apply(pd.Series)
````

````python
delays.rename(columns={0:'delay_1', 1:'delay_2', 2:'delay_3'}, inplace=True)
delays
````

````python
df = df.join(delays)
df.drop('RecentDelays', axis=1, inplace=True)
df
````

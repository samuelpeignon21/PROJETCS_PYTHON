### Enfin, parlons des heatmaps.

C'est un graphe très utile, car en analyse de données il est vital de repérer rapidement **les corrélations entre les variables dans l'EDA**.


Les heatmaps ou cartes de chaleurs sont particulièrement utiles pour visualiser les **corrélations entre variables**, car elles utilisent des variations de couleur pour montrer la force et la direction des relations entre elles.

Chaque cellule dans la **matrice de corrélation** est colorée selon le degré de corrélation, permettant d'identifier rapidement les relations positives, négatives ou nulles entre les différentes variables.

#### Cela facilite l'analyse des interconnexions au sein des données.


Tout graphique de corrélation se base sur un **intervalle de corrélation normalisé qui s'étend de -1 à 1,** où :

- 1 positif : indique une corrélation positive parfaite
- 1 négatif : indique une corrélation négative parfaite
- 0 aucune corrélation (relation neutre) entre les deux variables.

> _Les valeurs proches de 1 ou -1 signalent une forte corrélation, tandis que celles proches de 0 indiquent une faible ou inexistante relation linéaire._

#### **Quelques définitions pratiques :**


- **Corrélation positive :** Quand une variable augmente, l'autre augmente aussi.
- **Corrélation négative :** Quand une variable augmente, l'autre diminue.
- **Corrélation neutre :** Absence de relation linéaire significative entre deux variables.
- **Coefficient de corrélation :** Mesure la force et la direction de la relation linéaire entre deux variables.

```python
df_num = df[ ['checkout_price','base_price','num_orders'] ] # Que fait ce code ?
```

**Il nous retourne les valeurs numériques de notre jeu de données uniquement.**

#### Mais vous connaissez une façon plus efficiente et pérenne de le faire, n'est-ce pas ?

```python
display(df_num) # Visualisons les variables numériques.
```

```python
corrmat = df_num.corr() # La méthode '.corr()' génère une matrice de corrélation que l'on stocke dans une variable.

display(corrmat)
```

```python
# Le tuple renseigné dans le paramètre 'figsize=' de la méthode '.figure()' du package matplotlib permet de configurer les dimensions du graphique.
plt.figure(figsize=(8, 3))

sns.heatmap(corrmat, annot=True, fmt='.2%', cmap='magma', square=True)
plt.show()
```

**Que peut-on observer ?**

**`corrmat`** est notre **variable d'entrée** (matrice de corrélation), nous la passons dans la méthode `.heatmap()` pour générer la carte de chaleur.

Le paramètre **`annot=True`** affiche les **valeurs des coefficients de corrélation directement sur chaque cellule de la heatmap,** ce qui facilite leur lecture.

**`fmt=`** a pour but de **formater les annotations** (les valeurs affichées par le paramètre `annot=`), **en pourcentage en ajustant les décimales.**

Pour spécifier la **palette de couleurs** utilisée pour la heatmap, il faut paramétrer **`cmap=`**. ici `viridis`, une palette qui passe du bleu au jaune, souvent utilisée pour sa **perceptibilité uniforme.**

Et enfin, **`square=True`** **force les cellules** de la heatmap à **être carrées,** ce qui donne un aspect plus **harmonieux et facilite la comparaison visuelle entre les cellules.**


#### On voit, et cela semble logique, que la corrélation entre le prix de base et le prix final est très importante. Il est intéressant de voir la corrélation légèrement négative entre les variables de prix et le nombre de commandes.

### Comment choisir le type de graphique ?

Le type de graphique à utiliser dépend évidemment de nos données et de l'analyse qu'on veut mettre en place.

Les options principales :

* **les graphiques catégoriels (categorical plots)** (`sns.catplot()`): Lorsqu'on veut montrer nos données en se basant sur certaines **variables catégorielles**. Par exemple, le nombre de commandes par type de cuisine. L'axe y est numérique (nombre de commandes) et l'axe x est catégoriel (type de cuisine). Ce genre de graphique comprend les **diagrammes en bâtons (bar plot)**, **les boîtes à moustache (box plot)**, **les diagrammes en violon (violin plot)**. Le **camembert (pie plot)** est aussi catégoriel, mais pas directement faisable avec seaborn, on le fera avec matplotlib.

* **Les diagrammes de relation ou de corrélation (Relationship plots**  (`sns.replot()`): lorsqu'on veut montrer la relation entre **2 variables numériques**. Par exemple, le nombre de commandes vs la taille de la zone d'opération. Une variable est affichée sur l'axe x, l'autre sur l'axe y. Ce genre de graphique comprend les **Nuages de points (scatter plot)** et les **graphiques linéaires (line plot)**.



* **Les graphiques de distribution** (`sns.distplot()`) : quand on veut afficher la distribution **d'une seule variable numérique** par exemple en **histogramme**, pour voir comment le prix de la nourriture est distribué.



* **Les cartes de chaleur (heatmap)** (`sns.heatmap()`) : quand on veut montrer des corrélations entre les variables.

# Inférence statistique, bootstrapping et intervalles de confiance

### Rappelez-vous...

**L'inférence statistique** est la méthode par laquelle nous utilisons des données d'un échantillon pour tirer des conclusions sur une population plus large.

> Elle permet **d'estimer des paramètres inconnus** (comme la moyenne ou la proportion) à partir de l'échantillon et d'évaluer l'incertitude associée à ces estimations.

**Le bootstrapping** est une technique d'échantillonnage **aléatoire** répétée avec **remise** à partir des données observées.

> Elle permet de créer de **nombreux échantillons simulés** pour estimer la **variabilité** d'une statistique (comme la moyenne ou la médiane) sans faire d'hypothèses fortes sur la distribution des données.

Les **intervalles de confiance** sont des **plages** de valeurs dans lesquelles nous estimons que le paramètre de la **population** se trouve avec une **certaine probabilité** (généralement 95%).

> Ils quantifient **l'incertitude d'une estimation** et offrent un moyen de dire, par exemple, que la vraie moyenne est très probablement comprise dans **cet intervalle.**

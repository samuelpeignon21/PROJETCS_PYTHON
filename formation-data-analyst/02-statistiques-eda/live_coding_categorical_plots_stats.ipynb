## Importation des packages et fichiers.

**On commence par importer les packages.**


```python
import matplotlib.pyplot as plt
import seaborn as sns
import matplotlib

import pandas as pd
import numpy as np

from tqdm import tqdm

# On peut fixer des paramètres pour tout le notebook :

plt.style.use('seaborn-v0_8') # On configure le style graphique des graphs.
matplotlib.rcParams['figure.figsize'] = (15.0, 6.0) # On configure la taille des graphs.
```

#### Maintenant que nous avons chargé les packages et correctement initialisé le notebook, **il faut charger nos données.**

**Nous allons utiliser les données d'une entreprise de livraison de repas.**

- `df_meal` décrit le type de repas.
- `df_center` décrit le nom du centre d'approvisionnement.
- `df_food` décrit les commandes individuelles.

```python
df_meal = pd.read_csv("/content/meal_info.csv")
df_center = pd.read_csv("/content/fulfilment_center_info.csv")
df_food = pd.read_csv("/content/orders.csv")
```

```python
# 0n fusionne les informations en une seule DataFrame.

df = pd.merge(df_food, df_center, on='center_id') # Rappelez-vous, le paramètre 'on=' est utilisé pour définir la clé de jointure.
df = pd.merge(df, df_meal, on='meal_id') # Le paramètre 'how=' par défaut, est défini comme étant 'inner'.

display(df)
```

```python
# Préparons les données, on étudie le nombre de commandes, par type de cuisine et zone d'opération.

op_table = df.groupby(['op_area','cuisine'])['num_orders'].agg('sum').to_frame().reset_index() # Nous groupons les données par zones d'opération et types de cuisine, puis nous faisons la somme des commandes, que nous convertissons en dataframe avec index réinitialisé.

display(op_table)


# On peut aussi le faire avec la méthode pivot_table :
# op_table = pd.pivot_table(df, index = ['op_area','cuisine'], values = 'num_orders', aggfunc = np.sum).reset_index()
```

 Parlons des catégorical plots ou graphiques catégoriels.

Seaborn fournit la méthode  `.catplot()` pour afficher nos données par catégories.

Il permet de révéler des **patterns intéressants et de grouper nos données.**

> Les types de graphiques possibles (par l'argument `type`) sont : le graphe simple (par défaut, et pouvant être confondu avec le bar plot), le diagramme en bâtons, le diagramme en violon, et la boîte à moustache.

```python
# Ci-dessous dans les paramètres, nous renseignons la table préparée, nous spécifions les axes et le type du graphique.

sns.catplot(data=op_table, x='cuisine', y='num_orders', kind='bar', errorbar=None)
plt.show()
```

Comme on peut le voir, **on a juste besoin de dire à Seaborn où prendre les données, et choisir les données catégorielles à mettre en abscisse et les données numériques à mettre en ordonnée.** Ensuite, on peut ajouter d'autres arguments **si besoin,** comme `hue` et `swarm`.

```python
# 'hue' permet d'intégrer l'information d'une autre variable, ici la taille de la zone.

sns.catplot(data=op_table, x='cuisine', y='num_orders', hue='op_area') # Essayez le paramètre suivant 'kind='bar''.
plt.show()
```

#### Comme on peut le constater, `hue` permet de rajouter de l'information, plus précisement une **dimension** supplémentaire pour plus de **granularité** dans le graphique.

```python
# Essayons 'swarm'...
sns.catplot(data=op_table, x='cuisine', y='num_orders', hue='op_area', kind="swarm")
plt.show()
```

#### A quoi sert ce paramètre d'après vous ?*

### Passons aux **box-plot**.

Ce graphe que l'on vient de voir, n'est en fait rien de plus qu'un **boxplot non formaté.**

#### **On peut utiliser l'argument** `kind='box'` **pour adapter le format.**

```python
# On change d'argument dans le paramètre 'kind='.

sns.catplot(data=op_table, x='cuisine', y='num_orders', kind='box')
plt.show()
```

Un box plot ou diagramme en boîte (boîte à moustache encore pour les intimes) est un graphique qui représente la distribution d'un ensemble de données numériques à travers **ses quartiles, mettant en évidence la médiane, les quartiles, ainsi que les valeurs extrêmes et les éventuels points aberrants.**

**C'est justement la visualisation des points aberrants, également appelés `outliers`, qui nous intéresse.**

Il permet de visualiser rapidement la dispersion, la symétrie, et la concentration des données, tout en identifiant les valeurs qui se situent en dehors de la plage normale.

> Le box plot est crucial pour comparer des distributions entre plusieurs groupes ou catégories, **fournissant une vue d'ensemble claire des différences et similitudes entre elles.**

On pourrait aussi voir les prix en fonction du type de cuisine.

```python
sns.catplot(data=df, x='cuisine', y='base_price', kind='box') # Changeons l'ordonnée pour visualiser la répartition des prix de base.
plt.show()
```

### Un peu de styling avec matplotlib.

Pour ajuster les détails de notre graphique, **ajoutons quelques lignes code avec le package maplotlib dans une cellule.**

Par exemple, pour **réorienter les noms de lignes/colonnes** pour éviter la superposition, une méthode existe. Pour changer **la taille du graphe**, pour ajouter un plus gros **titre** au-dessus, on utilise toujours matplotlib.


```python
table2 = table.sort_values(by='num_orders', ascending=False).reset_index() # Reprenons notre table et réorganisons-la.
table2.drop(columns='index', inplace=True) # La colonne 'index' ne nous sert plus à rien, on peut la droper !

display(table2)
```

```python
# Même graphique avec les noms de catégories à la verticale, plus grands et avec un titre.

sns.catplot(data=table2, x='category', y='num_orders', kind="bar", height=6, aspect=2) # Pour l'instant, rien ne change.

# Mais nous pouvons rentrer dans la granularitée du graphe avec le package 'matplotlib' :

plt.xticks(rotation=30) # La méthode '.xticks()' accompagnée du paramètre 'rotation' permet de pivoter l'axe des abcisses.
plt.yticks(rotation=30) # Même chose avec l'axe des ordonnées.

plt.title('Nombre de commandes par catégorie de nourriture',  fontweight='bold', fontsize=24) # Nous rajoutons un titre...
plt.show()
```

**Tiens le graphique est ordonné de manière décroissante. Vous savez pourquoi ? Cherchez bien !**

#### **Pouvez-vous dire quels sont les éléments que prend la méthode `.pie()` ci-dessus ? À quoi sert le paramètre `labels=` ?**

```python
pie_data = df.groupby(['cuisine'])['num_orders'].sum()
print(pie_data) # On initialise une variable 'pie_data' pour pouvoir visualiser la répartition des commandes par type de cuisine.
```

```python
pie_data.index, pie_data.values # Vous vous rappelez des méthodes '.index' et '.values' ?
```

```python
plt.pie(pie_data.values, labels=pie_data.index)
plt.show() # Nous réutilisons la méthode '.pie()' avec notre pd.Series.
```

```python
# Ajoutons des options d'affichage :

plt.pie(pie_data.values,
        labels=pie_data.index,
        autopct='%0.1f%%',  # Permet d'afficher les pourcentages à un chiffre après la virgule.
        explode=[0, 0, 0.1, 0],   # Permet d'exploser les parts en fonction des entrées, de les faire ressortir.
        textprops={'weight': 'bold', 'fontsize': 14}) # Ce sont les propriétés des champs de texte.

plt.title('Cuisine share %', fontweight='bold', fontsize=14) # On ajoute un titre.
plt.show()
```

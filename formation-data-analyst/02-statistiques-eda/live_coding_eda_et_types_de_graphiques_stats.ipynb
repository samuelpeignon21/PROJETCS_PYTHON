# Découverte des graphiques Pandas.

Pandas fournit la méthode `.plot()` comme façon générale **d'afficher un dataframe sous forme de graphique.**

> Cette méthode est très versatile (une méthode pour différents graphiques, à l'aide de l'argument type) et accepte de nombreux arguments pour styliser le résultat.

**Cependant pour vraiment ajuster nos graphes à l'analyse, on a besoin de matplotlib.** D'ailleurs même pandas utilise matplotlib pour leur méthodes rapides de construction de graphiques.

#### Pratiquons donc la syntaxe de matplotlib (Pour en savoir plus sur les méthodes pandas : [clique ici](https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.DataFrame.plot.html)).

```python
df['checkout_price'] # Affichons la colonne des prix du dataframe.
```

#### Nous pouvons appliquer la méthode '.plot()' sur cette variable (la `pd.Series`) pour pouvoir la visualiser.

```python
df['checkout_price'].plot()
```

#### **Que constatez-vous ? Est-ce que le graphique ci-dessus est pertinent ?**

### Non, la donnée est trop condensée, il va falloir la traiter.

```python
df['checkout_price'][:200].plot() # Nous pouvons appliquer un slicing pour séléctionner les 200 premiers points de données.
```

#### Ici, par défaut, on constate que la méthode `.plot()` génère un **line chart.**

**Qu'est-ce qu'un line chart ?**

Un line chart ou graphique linéaire, est un type de graphique qui représente des données sous forme de points connectés par des lignes, **permettant de visualiser l'évolution d'une variable continue sur une période de temps ou selon une autre variable.**

> Il est particulièrement utile pour identifier des tendances, des cycles ou des variations dans les données.

**Un line chart est adapté si l'on veut visualiser l'évolution de données financières, comme dans notre cas.**

```python
df['cuisine'].value_counts() # Affichons la répartition des différentes cuisines de notre jeu de données.
```

```python
df['cuisine'].value_counts().plot(kind="bar") # Rajoutons simplement la méthode '.plot()', mais spécifions le paramètre 'kind'.
```

#### Qu'est-ce qu'on obtient ?

**Un bar chart ou graphique en barre,** est plus adapté pour visualiser des données catégorielles en chaînes de caractères car il permet de comparer facilement **les fréquences ou les totaux associés à chaque catégorie.**

> Les barres distinctes facilitent la comparaison visuelle des différentes catégories, mettant en évidence les différences et les distributions au sein des données.

```python
df['cuisine'].value_counts().plot(kind="pie") # Changeons l'argument du paramètre avec '.pie'.
```

#### Cette fois-ci, on obtient un **Pie Chart.**

Un pie chart ou diagramme circulaire (graphique en camembert pour les intimes), est super utile pour visualiser **la répartition proportionnelle des catégories au sein d'un ensemble de données.**

> Contrairement au bar chart, qui montre les valeurs absolues ou comparatives, le pie chart permet de voir immédiatement la **part relative de chaque catégorie** par rapport à l'ensemble, ce qui est idéal pour illustrer des pourcentages ou des parts de marché, par exemple.

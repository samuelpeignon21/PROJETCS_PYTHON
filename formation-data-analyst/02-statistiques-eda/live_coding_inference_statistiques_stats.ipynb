# Voyons maintenant le Bootstrapping, les inférences et les intervalles de confiance.


 #### **En data analyse, le bootstrapping est une technique de rééchantillonnage utilisée pour estimer la distribution d'un estimateur à partir de données limitées.**


En générant **plusieurs échantillons aléatoires** avec remplacement à partir des données originales, le bootstrapping permet de **calculer des intervalles de confiance** et d'obtenir des **mesures de la variabilité des estimateurs.**

> _Cela est particulièrement utile pour évaluer la **précision des estimations sans faire d'hypothèses strictes sur la distribution des données.** En utilisant NumPy, on peut facilement implémenter le bootstrapping, réaliser des inférences statistiques et estimer des intervalles de confiance pour les paramètres d'un modèle, offrant ainsi une **évaluation robuste des résultats et une meilleure compréhension des incertitudes associées aux analyses de données.**_

### Creusons la théorie avec un exemple, les oeufs de pâques :

Admettons que nous ayons reçu une très grande cargaison **d'oeufs de Pâques** et on veut déterminer le **poids moyen** de ces oeufs pour vérifier la **qualité de la marchandise**.

#### Pour cela on ne va pas peser l'ensemble de la cargaison mais simplement **10 oeufs.**

```python
eggs = [20, 20, 20, 20, 70, 70, 70, 50, 90, 80] # Commençons par initialiser une liste contenant le poids en gramme de nos 10 oeufs.
```

```python
# Visualisons-la directement avec un histogramme :

sns.histplot(x=eggs, bins=10) # Nous renseignons au paramètre 'x=' la variable 'eggs'.

plt.title('Distributions de l\'échantillon d\'Oeufs de Pâques.') # Ajoutons un titre.

plt.show() # Il ne nous reste plus qu'à afficher le graphique avec la méthode '.plot()'.
```

**Pourquoi visualiser notre liste ?**

Visualiser la liste nous permettra de voir comment s'articule la méthode du Bootstrapping.

### Ensuite, ajoutons un intervalle de confiance sur nos valeurs :

```python
np.mean(eggs)
```

```python
np.std(eggs)
```

```python
np.sqrt(len(eggs))
```

```python
mu = np.mean(eggs) # Nous stockons dans 'mu', la moyenne de notre liste de valeurs...

sigma = np.std(eggs) # ...Dans 'sigma' l'écart-type de notre liste de valeurs...

z = sigma / np.sqrt(len(eggs)) # ...Puis, dans 'z' l'écart-type divisé par la racine carrée de la longueur de l'échantillon de notre liste de valeurs, soit l'erreur standard.

print('Moyenne de la liste de valeurs :', mu, '\nEcart-type de la liste de valeurs :', sigma, '\nErreur standard arrondie de la liste de valeurs :', round(z, 1))
```

**Que faisons-nous au dessus :**

- `mu` représente la moyenne des valeurs de l'échantillon.

- `sigma` est l'écart-type, indiquant la variabilité des valeurs par rapport à la moyenne.

- `z` est l'erreur standard, fournissant une estimation de l'incertitude autour de la moyenne échantillonnale.


**Ces statistiques sont fondamentales pour construire des intervalles de confiance et effectuer des inférences statistiques.**

```python
# Calculons maintenant l'intervalle de confiance à 95%, en appliquant la formule mathématique adéquate :

# Ici, avant la virgule nous calculons le premier élément de l'intervalle, puis après la virgule, le deuxième...
intervalle_de_confiance = (mu - 1.96*z, mu + 1.96*z) # ...Soit l'erreur standard multipliée par '1.96' (pour un intervalle de confiance de 95%) puis soustraction/addition de cette valeur par la moyenne de l'échantillon.

print(intervalle_de_confiance)
```

**Pourquoi avons-nous appliqué cette formule ?**

Plus de précision sur l'application de la formule :

- `mu` est la moyenne de l'échantillon.

- `z` est l'erreur standard.

- `1.96` est le score z associé à un niveau de confiance de 95%.

> _Ce score provient de la distribution normale standard, et il représente le **nombre de déviations standard** nécessaires pour capturer **95%** des valeurs dans une distribution normale._

#### Pour d'autres niveaux de confiance, **les scores z sont différents :**

- `90%` : Le score `z` est environ `1.645`. Donc, l'intervalle de confiance est `(mu - 1.645*z, mu + 1.645*z)`.
- `99%` : Le score `z` est environ `2.576`. Donc, l'intervalle de confiance est `(mu - 2.576*z, mu + 2.576*z)`.

Ces valeurs sont tirées de la **distribution normale** et permettent de spécifier la **probabilité** que l'intervalle contienne la **véritable valeur du paramètre estimé.**

> _Plus le niveau de confiance est élevé, plus l'intervalle sera **large**, reflétant une **incertitude plus grande quant à la précision de notre estimation.**_

### Ci-dessous, la structure du Bootstrapping :

```python
bootstrap_means = [] # Initialisation d'une liste vide 'bootstrap_means' pour stocker les moyennes des échantillons bootstrappés.

# Exécution de la boucle pour créer 5000 échantillons bootstrappés.
for i in range(5000):
    # Génération d'un échantillon aléatoire avec remplacement de la taille de l'échantillon original.
    temp_sample = np.random.choice(eggs, replace=True, size=len(eggs))

    # Calcul de la moyenne de cet échantillon bootstrappé.
    temp_mean = np.mean(temp_sample)

    # Ajout de la moyenne calculée à la liste des moyennes bootstrappées.
    bootstrap_means.append(temp_mean)
```

```python
# Que faisons-nous exactement dans notre boucle ?

print(temp_sample) # Voici le résultat d'une itération avec génération aléatoire.
print(bootstrap_means) # Voici les 5000 moyennes générées à partir de nos dix oeufs.
```

```python
# Visualisons notre liste de 5000 moyennes :

sns.histplot(x=bootstrap_means, bins=25) # On rentre la  variable correspondante dans le paramètre 'x='.

plt.title('Echantillon bootstrappé des poids des oeufs.') # Ajoutons un titre.

plt.show() # Affichons le graphique.
```

```python
np.mean(bootstrap_means)
```

#### **_"Tout tend vers une distribution normale."_**


Lorsque qu'on **génère un grand nombre d'échantillons bootstrappés** et qu'on trace leur distribution, il est courant d'obtenir une forme qui ressemble à une **distribution normale**, même si les données **originales** ne sont pas normalement distribuées.

#### **Cela peut s'expliquer principalement par le Théorème de la Limite Centrale :**

> _Selon ce théorème, lorsque la **taille de l'échantillon** est **suffisamment grande**, la distribution des moyennes d'échantillons (ou d'autres statistiques similaires) **tend vers une distribution normale**, indépendamment de la forme de la distribution des **données d'origine**._

**En bootstrappant, on génère de nombreux échantillons et on calcule leurs moyennes, ce qui entraîne une distribution normale des moyennes, en raison de ce théorème.**

```python
# Nous pouvons calculer un intervalle de confiance basé sur notre distribution bootstrappée :

centiles = np.percentile(bootstrap_means, [2.5, 97.5])  # Calcul des percentiles 2.5% et 97.5%, que l'on stocke dans 'centiles'.
bootstrapped_ci = (round(centiles[0], 2), round(centiles[1], 2))  # Arrondir les valeurs pour plus de lisibilité.

print(bootstrapped_ci) # Affichons l'intervalle de confiance.
```

**Comment conclure ?**

L'intervalle de confiance de notre bootstrapping est `(34.0, 67.0)`, ce qui signifie que, selon les données que nous avons **échantillonnées et analysées**, on peut conclure que la **véritable moyenne** de la population sous-jacente se situe probablement entre `34.0` et `67.0` avec un **niveau de confiance** de 95%.

### Pratiquons avec un autre exemple : l'usine de chaussures.

> _Imaginons que nous possédions une **usine** qui **produit des chaussures**. Nous souhaitons être capables de **caractériser la taille moyenne des chaussures** et nous assurer qu'elles **répondent à certaines spécificités.**_

Notre usine produit des **centaines de chaussures** chaque jour, et il est impossible de **mesurer la taille de toutes les chaussures.**

#### Nous avons accès à un échantillon de `100` chaussures.

**Nous allons utiliser la méthode du bootstrap sur ces 100 tailles de chaussures pour obtenir un intervalle de confiance à 95% pour la taille moyenne de nos chaussures.**

Commençons par examiner la liste `shoe_lengths`, qui référence les 100 tailles de chaussures observées.

```python
# Ici on crée une liste qui initialise 100 tailles de chaussures.
# Un peu au hasard pour notre exemple.
shoe_lengths = np.random.normal(40, 1, 100) # Ici, on crée une liste qui initialise 100 tailles de chaussures.

print(shoe_lengths) # Affichons l'array.
```

```python
sns.histplot(shoe_lengths, bins=20) # Visualisons notre variable.
```

#### Déterminons maintenant nos intervalles de confiance théoriques.

```python
mu = np.mean(shoe_lengths) # Calculons la moyenne...

sigma = np.std(shoe_lengths) # ...L'écart-type...

z = sigma / np.sqrt(len(shoe_lengths)) # ...Puis enfin l'erreur standard.

print('Moyenne de la liste de valeurs :', mu, '\nEcart-type de la liste de valeurs :', sigma, '\nErreur standard arrondie de la liste de valeurs :', round(z, 1))

confidence_interval = (mu - 1.96*z, mu + 1.96*z) # Appliquons la formule pour générer un intervalle de confiance.
print(confidence_interval)
```

#### Intervalle de confiance bootstrappée.

Encore une fois, le bootstrapping consiste à **tirer pleins d'échantillons**  de taille `len(shoe_lengths)`,  de manière aléatoire avec remplacement, et ensuite calculer la **taille moyenne de chaque échantillon.**

#### Voici plus de détails sur le bootstrapping dans la cellule de code ci-dessous.

```python
# Initialisation d'une liste vide 'bootstrap_means' qui servira à stocker les moyennes des échantillons bootstrappés.
bootstrap_means = []

# Répétition de 1000 itérations pour générer 1000 échantillons bootstrappés.
for i in range(1000):
    # À chaque itération, un nouvel échantillon aléatoire avec remplacement est sélectionné à partir de 'shoe_lengths'...
    # ...Ayant la même taille que l'échantillon original. Le paramètre 'replace=True' permet d'échantillonner avec remplacement...
    # ...Ce qui signifie qu'un élément peut être sélectionné plusieurs fois dans le même échantillon.
    temp_sample = np.random.choice(shoe_lengths, replace=True, size=len(shoe_lengths))

    # Calcul de la moyenne des longueurs de chaussures pour cet échantillon bootstrappé.
    temp_mean = np.mean(temp_sample)

    # Ajout de la moyenne calculée à la liste 'bootstrap_means' qui stocke toutes les moyennes des échantillons générés.
    bootstrap_means.append(temp_mean)

print(bootstrap_means[:5])
print(len(bootstrap_means))
```

```python
sns.histplot(bootstrap_means, bins=20) # Visualisons chacune des 1000 moyennes générées aléatoirement.
```

Enfin, on calcule un intervalle de confiance de 95% bootstrappé (`boot_95_ci`) en utilisant `np.percentile()`.

```python
boot_95_ci = np.percentile(bootstrap_means, [2.5, 97.5])
print("95% Bootstrapped CI = {}".format(boot_95_ci))
```

Ici on en déduit que si on prend **n'importe quel échantillon**, il y a `95%` de chance que notre **taille moyenne** se trouve dans **l'intervalle de confiance**. <br>
Par extension, la **population totale** est un échantillon, donc **la vraie taille moyenne a `95%` de chance** de se trouver dans cet intervalle.

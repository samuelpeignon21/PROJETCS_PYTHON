#### Maintenant que nous avons chargé les packages et correctement initialisé le notebook, **il faut charger nos données.**

**Nous allons utiliser les données d'une entreprise de livraison de repas.**

- `df_meal` décrit le type de repas.
- `df_center` décrit le nom du centre d'approvisionnement.
- `df_food` décrit les commandes individuelles.

```python
df_meal = pd.read_csv("/content/meal_info.csv")
df_center = pd.read_csv("/content/fulfilment_center_info.csv")
df_food = pd.read_csv("/content/orders.csv")
```

```python
# 0n fusionne les informations en un seul dataframe :

df = pd.merge(df_food, df_center, on='center_id') # Rappelez-vous, le paramètre 'on=' est utilisé pour définir la clé de jointure.
df = pd.merge(df, df_meal, on='meal_id') # Le paramètre 'how=' par défaut, est défini comme étant 'inner'.

display(df)
```

### Parlons un peu des histogrammes.

Les histogrammes sont affichés par **Seaborn** avec `distplot()`, `histplot()` ou `displot()`.

**Pour le moment trois fonctions cohabitent, mais nous recommendons d'utiliser** `histplot()`.

Nous allons afficher la variable `base_price`, qui est une **variable numérique continue**, (type de variable indispensable pour faire un histogramme qui a du sens).

> On peut inclure une estimation de la **fonction de densité** avec l'argument `kde`, et ajuster le **nombre de subdivisions** avec l'argument `bins`.

```python
sns.histplot(df['base_price'], bins=10)
plt.show() # Jouer avec le paramètre 'bins' peut-être très utile pour visualiser la donnée sous différents angles, notamment pour détecter les anomalies.
```

```python
# avec KDE (Kernel Density Estimate) et moins de barres (bins)

sns.histplot(df['base_price'], kde=True, bins=20)
plt.show() # Le paramètre KDE (Kernel Density Estimate), trace la tendance de la distribution des données indépendemment du 'bins'.
```

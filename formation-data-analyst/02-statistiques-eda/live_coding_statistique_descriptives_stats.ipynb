# Les statistiques descriptives en Python.

**Les statistiques descriptives constituent un ensemble d’outils permettant de résumer et d’analyser les caractéristiques essentielles d’un jeu de données.**

> En Python, elles sont largement utilisées pour explorer, comprendre et visualiser les données avant de passer à des analyses plus complexes.


#### Ces statistiques incluent des mesures comme la **moyenne, la médiane, l’écart-type, ainsi que des distributions de fréquence.**

> Grâce à des bibliothèques comme `Pandas`, `NumPy` et `Matplotlib`, Python offre une **approche flexible et puissante** pour effectuer des analyses descriptives, facilitant ainsi la prise de décision et l’exploration des données.




**On commence par l'importation des packages et jeu de données :**

```python
import numpy as np
import scipy.stats as st # Le package 'scipy' ou scientific Python permet l'utilisation de méthodes statistiques.

import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
olist = pd.read_csv("/content/olist_orders_payments.csv")
olist
```

```python
olist.info() # La méthode '.info()' nous renvoie le détail des champs du DataFrame.
```

### Il est temps de faire de l'EDA...

Rappelons-nous de l’importance de l’EDA (Exploratory Data Analysis), une **étape cruciale** dans le processus **d’analyse de données**.

> Son principal objectif est de **comprendre la structure, les patterns et les relations** au sein des données.

- **Identifier les tendances principales** : telles que la répartition des valeurs, les tendances temporelles ou les variations géographiques.
- **Détecter les anomalies ou valeurs aberrantes** : qui pourraient biaiser les résultats d'analyses futures.
- **Vérifier les données manquantes** : et évaluer leur impact potentiel sur l’analyse.
- **Formuler des hypothèses** : sur les relations entre les variables, à tester lors d'analyses statistiques.
- **Choisir les méthodes d'analyse appropriées** : en fonction des caractéristiques observées.

```python
olist['payment_value'] = olist['payment_value'].str.replace(',', '.').astype(float) # Remplacement des virgules par des points dans la colonne payment_value.

# Puis, conversion de la colonne en type float pour permettre des opérations numériques.
```

```python
olist['payment_value'].describe() # Ce qui nous permet d'utiliser la méthode '.describe()'.
```

```python
olist['payment_value'].value_counts() # Quelle valeur est la plus représentée ?
```

```python
olist['payment_value'].isna()
```

```python
olist['payment_value'].isna().value_counts() # Y'a t-il des valeurs manquantes ?
```

```python
np.mean(olist['payment_value']) # La moyenne de la colonne.
```

```python
np.median(olist['payment_value']) # La médiane de la colonne.
```

```python
np.var(olist['payment_value']) # La variance de la colonne.
```

```python
st.mode(olist['payment_value']) # Le mode de la colonne (valeur la plus représentée).
```

```python
np.corrcoef(olist['payment_value'], olist['payment_installments']) # La corrélation entre les deux variables.
```

```python
# Nous affiche également une matrice mais plus jolie.
olist[["payment_value", "payment_installments"]].corr()
```

```python
olist['payment_value'].corr(olist['payment_installments']) # On utilise ici une autre méthode pour afficher uniquement la valeur qui nous intéresse.
```

#### Vous l'avez compris, les méthodes `.corr()` et `.corrcoef()` en Python sont utilisées pour **calculer les corrélations** entre variables dans un jeu de données.

- **`corr()`** provient du package **Pandas** : Cette méthode permet de calculer la matrice de corrélation entre toutes les colonnes d’un DataFrame. Elle prend en charge différents types de corrélations (comme Pearson, Kendall ou Spearman). La sortie est une matrice indiquant comment chaque variable est corrélée avec les autres.
- **`corrcoef()`** provient du package **NumPy** : Cette méthode renvoie également une matrice de corrélation, mais elle est plus générale et s’applique à des tableaux NumPy. Par défaut, elle calcule la corrélation de Pearson.

```python
np.corrcoef(olist["payment_value"], olist["payment_installments"])
```

**`sns.heatmap()`** permet de **visualiser une matrice de corrélation**, obtenue avec **`.corr()`** (Pandas) ou **`.corrcoef()`** (NumPy). `annot= True` permet de faire apparaître les valeurs sur le graphique.

```python
# Nous voulons maintenant un meilleur visuel pour cette matrice de corrélation.

sns.heatmap(np.corrcoef(olist["payment_value"], olist["payment_installments"]), annot=True)
plt.show()
# Petit rappel, le paramètre 'annot=' est utile car il nous affiche les coefficients de corrélation.
```

```python
# On range notre matrice de corrélation dans une variable.
corr_pay_val_inst = olist[["payment_value", "payment_installments"]].corr()
corr_pay_val_inst
```

```python
# Nous pouvons nous servir de la variable définie précédemment pour notre graphique heatmap.
sns.heatmap(corr_pay_val_inst, annot= True)
plt.show()
```
